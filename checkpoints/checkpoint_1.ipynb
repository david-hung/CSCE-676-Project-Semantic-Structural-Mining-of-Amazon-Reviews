# (B) Comparative Analysis of Datasets

| Dimension | PEMS-BAY Traffic | Online Retail II | Amazon Reviews |
|------------|------------------|------------------|-----------------|
| Supported Tasks | Graph mining, clustering, anomaly detection (course); GNN forecasting (external) | Frequent itemsets, clustering (course); sequential patterns (external) | Text mining, graph mining (course); topic modeling, transformers (external) |
| Data Quality Issues | Missing timestamps, sensor outages, non-stationarity | Missing CustomerID, cancellations, noisy descriptions | Noisy text, duplicates, review bias |
| Algorithmic Feasibility | Manageable size on single machine | Apriori heavy but feasible with threshold tuning | Very large; requires subsetting |
| Bias Considerations | Coverage bias (only instrumented roads) | Retailer + geographic bias | Self-selection bias in reviews |
| Ethical Considerations | Traffic surveillance implications | Customer profiling concerns | Risk of deanonymization or unfair targeting |



# (C) Final Dataset Selection: Amazon Reviews

## Reasons for Selection

1. Combines **text + graph + temporal structure**, making it the richest dataset.
2. Directly supports course topics:
   - Text Mining
   - Embeddings
   - Graph Mining
3. Provides clear external extension:
   - Transformer-based embeddings
   - Topic modeling
   - Graph neural networks
4. Flexible research directions:
   - Product communities
   - Review sentiment structure
   - Rating prediction
   - Fake review detection

## Trade-Offs

- Large dataset — must filter by:
  - Single category (e.g., Electronics)
  - Year range
  - Top-k products
- Requires cleaning noisy text
- Potential ethical risks if modeling users directly


# (D) Exploratory Data Analysis Plan (Amazon Dataset)

## 1. Data Basics

- Total number of reviews
- Number of unique users
- Number of unique products
- Review length distribution
- Rating distribution (1–5 stars)

## 2. Text Analysis

- Most common terms after cleaning
- Average review length by rating
- Word frequency by sentiment class

## 3. Graph Properties

Construct a bipartite graph:
- Nodes: Users + Products
- Edges: Review interactions

Measure:
- Degree distribution
- Connected components
- Sparsity
- Power-law behavior

Optional:
- Product–product co-review graph
- Community detection results

## 4. Temporal Analysis

- Reviews per year/month
- Rating trends over time
- Burst detection

## 5. Data Cleaning Strategy (Explain WHY)

- Remove duplicate reviews
- Remove empty or extremely short text
- Lowercase + tokenize
- Remove stopwords
- Handle rare users/products



# (E) Initial Research Direction

## Observations to Investigate

- Ratings likely heavily skewed toward 4–5 stars
- A small number of products dominate review volume
- Graph expected to follow power-law degree distribution

## Hypotheses

- Transformer embeddings will produce clearer product clusters than TF-IDF.
- Products within the same learned topic cluster will share similar rating distributions.
- Community detection on product graph will align with semantic themes.

## Potential Research Questions

1. Do transformer-based embeddings improve product similarity over traditional methods?
2. Does graph structure reveal product communities not visible in text alone?
3. Can topic modeling identify latent themes associated with negative reviews?
4. Are highly connected users more likely to give extreme ratings?
