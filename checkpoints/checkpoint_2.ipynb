## EDA Summary

Initial exploratory analysis revealed several key properties:

- **Rating Distribution:** Highly skewed toward 4–5 stars → limited variance in labels  
- **Text Properties:** Reviews are noisy, with large variation in length and vocabulary  
- **Long-Tail Distribution:** A small number of products dominate review counts  
- **Graph Structure:** User–product interactions form a sparse bipartite graph with power-law degree distribution  

These findings suggest that:
- Product similarity is non-trivial (text vs structure may differ)
- Clustering quality may be sensitive to representation choice
- Additional methods beyond clustering may reveal deeper patterns

## RQ1 (Course)
**How do product clusters differ when using text-based representations (TF-IDF) versus graph-based representations (node2vec)?**

- **Task Type:** Clustering + Representation Learning  
- **Algorithms:**
  - TF-IDF + KMeans (course)
  - node2vec + KMeans (course)
- **Evaluation Criteria:**
  - Silhouette score
  - Cluster interpretability
  - Cluster consistency

---

## RQ2 (Course)
**Does graph structure (e.g., centrality or connectivity) correlate with product rating behavior?**

- **Task Type:** Graph Mining + Predictive Analysis  
- **Algorithms:**
  - Degree centrality / PageRank (course)
  - Regression or classification models (course ML)
- **Evaluation Criteria:**
  - Correlation coefficients
  - Predictive performance (RMSE / accuracy)

## RQ3 (External)
**Can topic modeling uncover latent product themes that are not captured by clustering methods?**

- **Task Type:** Topic Modeling  
- **Algorithms:**
  - TF-IDF + KMeans (baseline, course)
  - LDA or BERTopic (external)

# (4) Motivation and Feasibility

## Motivation
- Text captures semantic similarity  
- Graph captures interaction-based relationships  
- It is unclear which better reflects true product similarity  

## Non-Triviality
- Comparing fundamentally different representations (text vs graph)
- Clustering may miss deeper latent themes
- Topic modeling introduces a probabilistic structure

## Feasibility
- TF-IDF and KMeans are computationally efficient
- node2vec is scalable for medium-sized graphs
- LDA is widely available in standard libraries

## Risks
- Sparse graph may reduce embedding quality  
- Topic modeling sensitive to parameter choice  
- Noisy text may reduce interpretability

# (5.1) Algorithm Justification

- **TF-IDF:**  
  Chosen as a baseline text representation because it is simple, interpretable, and effective for sparse high-dimensional text data.

- **KMeans:**  
  Selected for its efficiency and scalability, allowing consistent clustering across different representations.

- **node2vec:**  
  Used to capture structural relationships in the graph through random walks, enabling comparison with text-based similarity.

- **Centrality Measures (PageRank / Degree):**  
  Provide interpretable measures of node importance, allowing analysis of how product position in the graph relates to ratings.

- **Regression / Classification Models:**  
  Used to evaluate predictive relationships between graph features and ratings while maintaining interpretability.

- **LDA / BERTopic (External):**  
  Selected to uncover latent semantic themes that clustering methods may miss, introducing a different modeling approach.
